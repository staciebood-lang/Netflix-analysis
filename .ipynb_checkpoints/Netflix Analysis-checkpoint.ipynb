```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/netflix_titles.csv")

df.head()
```

```python
df["country"] = df["country"].fillna("Unknown")
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["date_added"] = df["date_added"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Unknown")
df["duration"] = df["duration"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["type"].value_counts()
```

```python
df["type"].value_counts().plot(kind="bar")

plt.title("Netflix Movies vs. TV Shows")
plt.ylabel("Count")
plt.show()
```

```python
df["country"].value_counts().head(10)
```

```python
df["country"].value_counts().head(10).plot(kind = "barh")

plt.title("Top 10 Countries for Netflix Content")
plt.ylabel("Country")
plt.show()
```

```python
df["release_year"].value_counts().plot(kind = "line")
plt.ylabel = "Number of movies released"
```

## Key Insights

1. Netflix has more Movies than TV Shows.
2. Most content comes from a small number of countries.
3. Content production increased significantly after 2010.
